```python
## Step 14: Markdown Justification
## Missing Data Handling Strategy
'''
The missing values were handled based on the nature and importance
of each column.

- Age: Median imputation was selected because age is numerical and
  the median is less sensitive to extreme values.
- Gender: Mode imputation was selected because Gender is categorical.
- Product Category: Mode imputation was used because it is categorical.
- Quantity: Median imputation was used because Quantity is numerical.
- Price per Unit: Median imputation was used to reduce the influence
  of extreme values.
- Transaction ID, Customer ID and Date: Rows with missing values
  in these critical fields were removed because these fields are
  required for identifying and analysing transactions.
  '''
```

```python
## Step 15: Handle missing values
```

```python
df = df.dropna(
    subset=['Transaction ID', 'Customer ID', 'Date']
)
```

```python
## Gender -> Mode
df['Gender'] = df['Gender'].fillna(
    df['Gender'].mode()[0]
)
```

```python
## Age -> Median
df['Age'] = df['Age'].fillna(
    df['Age'].median()
)
```

```python
## Product Category -> Mode
df['Product Category'] = df['Product Category'].fillna(
    df['Product Category'].mode()[0]
)
```

```python
## Quantity -> Median
df['Quantity'] = df['Quantity'].fillna(
    df['Quantity'].median()
)
```

```python
## Price per Unit -> Median
df['Price per Unit'] = df['Price per Unit'].fillna(
    df['Price per Unit'].median()
)
```

```python
## Total Amount
df['Total Amount'] = df['Total Amount'].fillna(
    df['Quantity'] * df['Price per Unit']
)
```

```python
## Step 19: Standardisation - Gender
df['Gender'].value_counts(dropna=False)
```

```python
df['Gender'] = df['Gender'].astype(str).str.strip().str.lower()

df['Gender'] = df['Gender'].replace({
    'm': 'Male',
    'male': 'Male',
    'f': 'Female',
    'female': 'Female'
})
```

```python
## Step 20: Standardise Product Category
print(df['Product Category'].unique())

## Remove extry space
df['Product Category'] = (
    df['Product Category']
    .astype(str)
    .str.strip()
)
```

```python
df['Product Category'] = (
    df['Product Category']
    .str.title()
)
df['Product Category'].unique()
```

```python
## Step 21: Standardise Date
print(df['Date'].head(10))
print(df['Date'].dtype)
df['Date'] = pd.to_datetime(
    df['Date'],
    errors='coerce'
)
df['Date'].isnull().sum()
df['Date'].isnull().sum()
```

```python
## Step 22: Standardise IDs
df['Transaction ID'] = (
    df['Transaction ID']
    .astype(str)
    .str.strip()
)

df['Customer ID'] = (
    df['Customer ID']
    .astype(str)
    .str.strip()
)
```

```python
## Step 23: Correct Numerical Data types
## Convert Age:
df['Age'] = pd.to_numeric(
    df['Age'],
    errors='coerce'
)

## Quantity
df['Quantity'] = pd.to_numeric(
    df['Quantity'],
    errors='coerce'
)

## Price
df['Price per Unit'] = pd.to_numeric(
    df['Price per Unit'],
    errors='coerce'
)

## Total Amount
df['Total Amount'] = pd.to_numeric(
    df['Total Amount'],
    errors='coerce'
)
```

```python
## Step 24: Check Data Types Again
df.dtypes
```

```python
## Step 25: Outlier Detection Using IQR
Q1 = df['Age'].quantile(0.25)
Q3 = df['Age'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
age_outliers = df[
    (df['Age'] < lower_bound) |
    (df['Age'] > upper_bound)
]

print("Age outliers:", len(age_outliers))
```

```python
## Step 26: Create a Reusable IQR Function
def detect_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = data[
        (data[column] < lower) |
        (data[column] > upper)
    ]

    return outliers, lower, upper
```

```python
numeric_columns = [
    'Age',
    'Quantity',
    'Price per Unit',
    'Total Amount'
]

for column in numeric_columns:
    outliers, lower, upper = detect_outliers_iqr(df, column)

    print("\n", column)
    print("Lower Bound:", lower)
    print("Upper Bound:", upper)
    print("Outliers:", len(outliers))
```

```python
## Step 27: Handle Impossible Age Values
invalid_age = df[
    (df['Age'] < 0) |
    (df['Age'] > 100)
]

print("Invalid age records:", len(invalid_age))

## remove them:
df = df[
    (df['Age'] >= 0) &
    (df['Age'] <= 100)
]
```

```python
print(df[df['Price per Unit'] < 0])
df = df[df['Price per Unit'] >= 0]
```

```python
## Step 28: Document Outlier Decision
## Outlier Handling Strategy
'''
The IQR method was used to identify statistical outliers in
Age, Quantity, Price per Unit, and Total Amount.

Not every statistical outlier was removed because extreme values
can represent legitimate retail transactions. For example, a
high-value transaction may be a genuine purchase.

Clearly invalid business values, such as negative age or negative
price, were removed. Legitimate high-value transactions were retained
to avoid losing useful business information.
'''
```
